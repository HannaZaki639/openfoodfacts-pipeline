```python
import pandas as pd
import json
```

```python
with open ("../data/raw_data.json","r", encoding = "utf-8") as file:
    raw_data = json.load(file)
    
```

```python
raw_data
```

```python
'''loop over each product pop out nutriments 
append product to its new place without nutriments
loop over nutriments to extract key value and add code (product id) to it 
save both to dataframe'''
```

```python
products =[]
nutriments =[]
for product in raw_data:
    nutri =product.pop('nutriments',{})
    products.append(product)
    
    for key,value in nutri.items():
        nutriments.append({'code': product['code'],
                           'nutriments':key,
                           'value':value})
    
    
```

```python
products[:2]
```

```python
nutriments_df = pd.DataFrame(nutriments)
nutriments_df
```
